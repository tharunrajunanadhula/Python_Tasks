```python
# 1. Print Hello World
if __name__ == '__main__':
    message = "Hello, World!"
    print(message)
```

```python
# 2. Check Weird or Not Weird
if __name__ == '__main__':
    number = int(input().strip())

    if number % 2 != 0:
        print("Weird")
    else:
        if 2 <= number <= 5:
            print("Not Weird")
        elif 6 <= number <= 20:
            print("Weird")
        else:
            print("Not Weird")
```

```python
# 3. Arithmetic Operations
if __name__ == '__main__':
    num1 = int(input())
    num2 = int(input())

    print(num1 + num2)
    print(num1 - num2)
    print(num1 * num2)
```

```python
# 4. Division Output
if __name__ == '__main__':
    first = int(input())
    second = int(input())

    print(first // second)
    print(first / second)
```

```python
# 5. Print Squares
if __name__ == '__main__':
    limit = int(input())

    for value in range(limit):
        print(value ** 2)
```

```python
# 6. Leap Year Function
def is_leap(year_value):
    leap_flag = False

    if (year_value % 4 == 0 and year_value % 100 != 0) or (year_value % 400 == 0):
        leap_flag = True

    return leap_flag
```

```python
# 7. Create Incremental Number Output
if __name__ == '__main__':
    total = int(input())

    output = 0
    for num in range(1, total + 1):
        output = output * 10 + num
    print(output)
```
