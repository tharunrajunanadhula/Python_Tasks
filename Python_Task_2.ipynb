```python
# Assignment - 2

from typing import List

# 1. Running Sum of an Array
class Solution:
    def runningSum(self, values: List[int]) -> List[int]:
        total = 0
        result_list = []
        for num in values:
            total += num
            result_list.append(total)
        return result_list
```

```python
# 2. Shuffle the Array
class Solution:
    def shuffle(self, items: List[int], split_point: int) -> List[int]:
        output = []
        for index in range(split_point):
            output.append(items[index])
            output.append(items[index + split_point])
        return output
```

```python
# 3. Kids With Greatest Number of Candies
class Solution:
    def kidsWithCandies(self, candy_list: List[int], bonus: int) -> List[bool]:
        highest = max(candy_list)
        outcome = []
        for amount in candy_list:
            outcome.append(amount + bonus >= highest)
        return outcome
```
